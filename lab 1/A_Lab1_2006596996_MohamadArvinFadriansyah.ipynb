## Explore Data

**2. Lakukan *exploratory data analysis* dengan menjawab pertanyaan-pertanyaan berikut!**

**a. Bagaimana sebaran jumlah guru di berbagai wilayah DKJ?**

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Mengelompokkan data berdasarkan wilayah dan menghitung total guru
guru_by_wilayah = df.groupby('wilayah')['jumlah_guru'].sum().reset_index()

# Visualisasi dengan bar plot
plt.figure(figsize=(20,10))
sns.barplot(data=guru_by_wilayah, x='wilayah', y='jumlah_guru', palette='viridis')
plt.xticks(rotation=45)
plt.xlabel('Wilayah')
plt.ylabel('Jumlah Guru')
plt.title('Sebaran Jumlah Guru di Berbagai Wilayah DKJ')
plt.show()
```

**b. Bagaimana rasio Guru-Murid di setiap jenjang pendidikan?**

```python
# Menambahkan kolom rasio guru-murid
df['rasio_guru_murid'] = df['jumlah_guru'] / df['jumlah_peserta_didik']

# Menghitung rata-rata rasio per jenjang pendidikan
rasio_by_jenjang = df.groupby('jenjang')['rasio_guru_murid'].mean().reset_index()

# Visualisasi
plt.figure(figsize=(8,6))
sns.barplot(data=rasio_by_jenjang, x='jenjang', y='rasio_guru_murid', palette='magma')
plt.xlabel('Jenjang Pendidikan')
plt.ylabel('Rata-rata Rasio Guru-Murid')
plt.title('Rasio Guru-Murid per Jenjang Pendidikan')
plt.show()
```

**c. Bagaimana distribusi jumlah sekolah di setiap wilayah?**

```python
# Mengelompokkan data berdasarkan wilayah untuk menghitung jumlah sekolah
sekolah_by_wilayah = df.groupby('wilayah')['jumlah_sekolah'].sum().reset_index()

# Visualisasi dengan bar plot
plt.figure(figsize=(20,10))
sns.barplot(data=sekolah_by_wilayah, x='wilayah', y='jumlah_sekolah', palette='coolwarm')
plt.xticks(rotation=45)
plt.xlabel('Wilayah')
plt.ylabel('Jumlah Sekolah')
plt.title('Distribusi Jumlah Sekolah di Setiap Wilayah')
plt.show()
```

**d. Apakah ada korelasi antara fitur (numerikal dan kategorikal) dengan target? Apakah fitur yang berkorelasi kuat tersebut juga saling berkorelasi?**

```python
# Ideal ratio is 1:20 (0.05)
tolerance = 0.05

# Assigning target based on ratio threshold
df['target'] = df['rasio_guru_murid'].apply(lambda x: 1 if x >= tolerance else 0)
```

```python
# Melakukan one-hot encoding pada fitur kategorikal
data_encoded = pd.get_dummies(df, columns=['wilayah', 'status_sekolah'])
# Membuat mapping manual untuk ordinal encoding jenjang pendidikan
jenjang_mapping = {
    'PAUD': 0,
    'PKBM': 1,
    'SD': 2,
    'SMP': 3,
    'SMA': 4,
    'SMK': 5,
    'SLB': 6
}

# Mengubah kolom 'jenjang' menjadi angka berdasarkan mapping
data_encoded['jenjang'] = data_encoded['jenjang'].map(jenjang_mapping)
data_encoded.head(20)
```

```python
# Menghitung matriks korelasi
corr_matrix = data_encoded.corr()

target_corr = corr_matrix['target'].sort_values(ascending=False)
print("Korelasi dengan target:")
print(target_corr)

# Visualisasi matriks korelasi menggunakan heatmap
plt.figure(figsize=(15,10))
sns.heatmap(corr_matrix, annot=True, fmt=".2f", cmap='coolwarm')
plt.title("Correlation Heatmap")
plt.show()
```

## Generate Test Design

Tim *data analyst* pribadi Pak Mindat memberikan Anda dua fungsi untuk melakukan *splitting*. Anda diminta menerapkan kedua fungsi tersebut, membandingkan pengaruhnya terhadap model, dan menentukan fungsi mana yang merupakan *good practice* dalam *machine learning*.

```python
def split_data_1(df, target_column, test_size=0.2, random_state=42):
    X = df.drop(columns=[target_column])
    y = df[target_column]

    X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=test_size, random_state=random_state)

    scaler = StandardScaler()
    X_train_scaled = scaler.fit_transform(X_train)
    X_test_scaled = scaler.transform(X_test)

    return X_train_scaled, X_test_scaled, y_train, y_test
```

```python
def split_data_2(df, target_column, test_size=0.2, random_state=42):
    X = df.drop(columns=[target_column])
    y = df[target_column]

    scaler = StandardScaler()
    X_scaled = scaler.fit_transform(X)

    X_train, X_test, y_train, y_test = train_test_split(X_scaled, y, test_size=test_size, random_state=random_state)

    return X_train, X_test, y_train, y_test
```

**4b. Lakukan *splitting* dataset menggunakan kedua fungsi sehingga Anda mendapat dua pasang data *training* dan *testing*!**

```python
# Penerapan split_data_1 (splitting dilakukan terlebih dahulu, kemudian scaling)
X_train_scaled_1, X_test_scaled_1, y_train_1, y_test_1 = split_data_1(df, target_column='target', test_size=0.2, random_state=42)

# Penerapan split_data_2 (scaling dilakukan pada seluruh data, kemudian splitting)
X_train_scaled_2, X_test_scaled_2, y_train_2, y_test_2 = split_data_2(df, target_column='target', test_size=0.2, random_state=42)

# Menampilkan dimensi masing-masing dataset untuk verifikasi
print("Menggunakan split_data_1:")
print("X_train_scaled_1 shape:", X_train_scaled_1.shape)
print("X_test_scaled_1 shape:", X_test_scaled_1.shape)
print("y_train_1 shape:", y_train_1.shape)
print("y_test_1 shape:", y_test_1.shape)

print("\nMenggunakan split_data_2:")
print("X_train_scaled_2 shape:", X_train_scaled_2.shape)
print("X_test_scaled_2 shape:", X_test_scaled_2.shape)
print("y_train_2 shape:", y_train_2.shape)
print("y_test_2 shape:", y_test_2.shape)
```

## Build Model

**4c. Buat dan latihlah model-model yang Anda sebutkan di nomor 4a! Masing-masing model harus menggunakan kedua hasil *split* data!** *(Misalnya, jika Anda memilih dua model, maka terdapat empat model yang dilatih)*

```python
from sklearn.linear_model import LinearRegression
from sklearn.ensemble import RandomForestRegressor
# ===============================================
# 1. Linear Regression
# ===============================================

# Model Linear Regression dengan hasil split_data_1
lr_model_1 = LinearRegression()
lr_model_1.fit(X_train_scaled_1, y_train_1)
y_pred_lr_1 = lr_model_1.predict(X_test_scaled_1)

# Model Linear Regression dengan hasil split_data_2
lr_model_2 = LinearRegression()
lr_model_2.fit(X_train_scaled_2, y_train_2)
y_pred_lr_2 = lr_model_2.predict(X_test_scaled_2)

# ===============================================
# 2. Random Forest Regression
# ===============================================

# Model Random Forest dengan hasil split_data_1
rf_model_1 = RandomForestRegressor(random_state=42)
rf_model_1.fit(X_train_scaled_1, y_train_1)
y_pred_rf_1 = rf_model_1.predict(X_test_scaled_1)


# Model Random Forest dengan hasil split_data_2
rf_model_2 = RandomForestRegressor(random_state=42)
rf_model_2.fit(X_train_scaled_2, y_train_2)
y_pred_rf_2 = rf_model_2.predict(X_test_scaled_2)

```
